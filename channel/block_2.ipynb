```python
import math

# 1. Таблица 1: Триплеты с миссмэтчем C-T/C-A/C-C (dG_37 в ккал/моль)
# Формат: верхняя нить 5'-XYZ-3' -> нам важен только сам триплет XYZ
MISMATCH_TRIMERS = {
    "ACC": 1.62, "CCC": 1.60, "GCA": 1.37, "GCC": 1.60,
    "GCG": 1.02, "GCT": 1.35, "TCC": 1.95, "ACA": 1.39,
    "ACG": 1.04, "ACT": 1.37, "CCA": 1.37, "CCG": 1.02,
    "CCT": 1.35, "TCA": 1.72, "TCG": 1.37, "TCT": 1.70
}

# 2. Таблица 2: Идеальные Уотсон-Криковские дуплеты (Unified, ккал/моль)
# Сразу разворачиваем симметрию на все 16 возможных динуклеотидов
WC_DOUBLETS = {
    "AA": -1.00, "TT": -1.00,
    "AT": -0.88,
    "TA": -0.58,
    "CA": -1.45, "TG": -1.45,
    "GT": -1.44, "AC": -1.44,
    "CT": -1.28, "AG": -1.28,
    "GA": -1.30, "TC": -1.30,
    "CG": -2.17,
    "GC": -2.24,
    "GG": -1.84, "CC": -1.84
}

RT = 0.616  # RT при 37 C (310.15 K) в ккал/моль

results = []

# 3. Основной цикл расчета
for trimer, dg_mismatch in MISMATCH_TRIMERS.items():
    # Разбиваем триплет XYZ на два шага: XY и YZ
    step1 = trimer[0:2]
    step2 = trimer[1:3]
    
    # Считаем энергию идеального комплементарного триплета
    dg_correct = WC_DOUBLETS[step1] + WC_DOUBLETS[step2]
    
    # Находим цену ошибки (Delta Delta G)
    ddg = (dg_mismatch - dg_correct)* 4184 / (6.02 * 10**23)
    
    # Считаем вероятность ошибки по Больцману
    const = 1/(1.380649*(10**(-23))*273)
    p_error = math.exp(-ddg * const)
    
    results.append({
        "trimer": trimer,
        "dg_mismatch": dg_mismatch,
        "dg_correct": dg_correct,
        "ddg": ddg,
        "p_error": p_error
    })

# 4. Сортируем по вероятности ошибки (от минимальной к максимальной)
results_sorted = sorted(results, key=lambda x: x["p_error"])

min_case = results_sorted[0]
max_case = results_sorted[-1]
mean_error = np.mean(results_sorted)

print("=== МИНИМАЛЬНАЯ ВЕРОЯТНОСТЬ ОШИБКИ ===")
print(f"Триплет: {min_case['trimer']} | dG_mismatch = {min_case['dg_mismatch']} | dG_correct = {min_case['dg_correct']:.2f}")
print(f"Delta Delta G = {min_case['ddg']} | P_error = {min_case['p_error']:.2e}\n")

print("=== МАКСИМАЛЬНАЯ ВЕРОЯТНОСТЬ ОШИБКИ ===")
print(f"Триплет: {max_case['trimer']} | dG_mismatch = {max_case['dg_mismatch']} | dG_correct = {max_case['dg_correct']:.2f}")
print(f"Delta Delta G = {max_case['ddg']} | P_error = {max_case['p_error']:.2e}")
```
